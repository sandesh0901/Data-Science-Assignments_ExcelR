```python
import pandas as pd

# Load datasets
patient_df = pd.read_csv("Patient_Data.csv")
billing_df = pd.read_csv("Billing_Data.csv")

# Display the datasets
print("Patient Dataset:")
display(patient_df)

print("\nBilling Dataset:")
display(billing_df)
```

```python
# Merge patient and billing datasets using PatientID
merged_df = pd.merge(
    patient_df,
    billing_df,
    on="PatientID",
    how="left"
)

print("Merged Dataset:")
display(merged_df)
```

```python
# New patient records
new_patient_records = pd.DataFrame({
    "PatientID": [107, 108],
    "Name": ["George", "Helen"],
    "Department": ["Cardiology", "Neurology"],
    "Doctor": ["Dr. Smith", "Dr. John"],
    "BillAmount": [5800, 6900],
    "ReceptionistID": [2, 3],
    "CheckInTime": ["2023-01-16 09:30", "2023-01-17 10:15"]
})

# Concatenate row-wise
patient_updated = pd.concat(
    [patient_df, new_patient_records],
    ignore_index=True
)

print("Patient Dataset after adding new records:")
display(patient_updated)
```

```python
# New billing category columns
new_billing_categories = pd.DataFrame({
    "BillingCategory": [
        "Premium",
        "Standard",
        "Premium",
        "Standard",
        "Basic"
    ],
    "PaymentStatus": [
        "Paid",
        "Paid",
        "Pending",
        "Paid",
        "Pending"
    ]
})

# Concatenate columns-wise
billing_updated = pd.concat(
    [billing_df, new_billing_categories],
    axis=1
)

print("Billing Dataset after adding new columns:")
display(billing_updated)
```
